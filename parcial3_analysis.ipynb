### 1. Importar Librerías y Módulos

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Importar módulos del proyecto
from data_analysis import DataAnalyzer
from preprocessing import DataPreprocessor
from models import MLModelTrainer
from evaluation import ModelEvaluator
from visualizations import MLVisualizer

# Configuración de visualización
plt.style.use('default')
plt.rcParams['figure.figsize'] = (12, 8)
sns.set_palette("husl")

# Mostrar todas las columnas
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

print("Librerías importadas exitosamente!")
```

### 2. Generar o Cargar Datos

```python
# Generar datos de muestra para demostración
analyzer = DataAnalyzer(pd.DataFrame())
data = analyzer.generate_sample_data(1000)

print(f"Datos generados: {data.shape}")
print("\nPrimeras 5 filas:")
data.head()
```

### 3. Análisis Exploratorio de Datos

```python
# Crear analizador
analyzer = DataAnalyzer(data)

# Información básica
basic_info = analyzer.basic_info()
```

### 5. Preprocesamiento de Datos

```python
# Crear preprocesador
preprocessor = DataPreprocessor(data, target_column='purchase')

# Aplicar preprocesamiento
preprocessor.handle_missing_values() \
           .remove_outliers(method='iqr', threshold=2.0) \
           .encode_categorical_variables(method='onehot') \
           .create_feature_interactions(max_interactions=3) \
           .scale_features(method='standard')

# Dividir datos
X_train, X_val, X_test, y_train, y_val, y_test = preprocessor.split_data()

print(f"Datos de entrenamiento: {X_train.shape}")
print(f"Datos de validación: {X_val.shape}")
print(f"Datos de prueba: {X_test.shape}")
```

### 6. Entrenamiento de Modelos

```python
# Crear entrenador de modelos
trainer = MLModelTrainer(problem_type='classification')

# Entrenar algunos modelos seleccionados
models_to_train = ['Logistic Regression', 'Random Forest', 'Gradient Boosting', 'SVM']

for model_name in models_to_train:
    print(f"\nEntrenando {model_name}...")
    trainer.train_single_model(model_name, X_train, y_train, optimize_hyperparams=False)

# Mostrar resumen de entrenamiento
results_df = trainer.print_training_summary()
```

### 7. Evaluación de Modelos

```python
# Generar predicciones
predictions = trainer.predict_with_all_models(X_test)
probabilities = trainer.predict_proba_with_all_models(X_test)

# Crear evaluador
evaluator = ModelEvaluator(problem_type='classification')

# Evaluar todos los modelos
comparison_df = evaluator.evaluate_multiple_models(y_test, predictions, probabilities)

print("\nResultados de evaluación:")
comparison_df
```

### 8. Visualizaciones de Resultados

```python
# Comparación de modelos
visualizer.plot_model_comparison(comparison_df)
```

```python
# Matrices de confusión
visualizer.plot_confusion_matrices(y_test, predictions)
```

```python
# Curvas ROC
visualizer.plot_roc_curves(y_test, probabilities)
```

```python
# Curvas Precision-Recall
visualizer.plot_precision_recall_curves(y_test, probabilities)
```

```python
# Importancia de características
feature_names = X_train.columns.tolist()
feature_importance = trainer.get_feature_importance(feature_names)

if feature_importance:
    visualizer.plot_feature_importance(feature_importance)
```

### 9. Análisis Detallado del Mejor Modelo

```python
# Obtener el mejor modelo
best_model_name = comparison_df.iloc[0]['Model']
print(f"Mejor modelo: {best_model_name}")

# Reporte detallado
detailed_report = evaluator.detailed_classification_report(best_model_name)
```

```python
# Análisis de umbrales
if best_model_name in probabilities:
    threshold_analysis = evaluator.threshold_analysis(
        y_test, probabilities[best_model_name], best_model_name
    )
```
